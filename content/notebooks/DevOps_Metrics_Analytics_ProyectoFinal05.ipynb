```python
import pandas as pd
df = pd.read_csv('/content/datos/dataset_limpio.csv')
print(df.info())
print(df.head())
print(df.isnull().sum())
```

```python
print("incidentes con resolution time = nulo vs incident_count:")
print(df[df['avg_resolution_time_hours'].isnull()]['incident_count'].value_range if hasattr(df['incident_count'], 'value_range') else df[df['avg_resolution_time_hours'].isnull()]['incident_count'].value_counts())
```

```python
print(df['avg_availability_pct'].describe())
```

```python
df_clean = df.copy()

# convertir datos donde  deployment_count == 0 or null
df_clean['avg_lead_time_hours'] = df_clean['avg_lead_time_hours'].fillna(0)
df_clean['deployment_failure_rate_pct'] = df_clean['deployment_failure_rate_pct'].fillna(0)
df_clean['rollback_rate_pct'] = df_clean['rollback_rate_pct'].fillna(0)

# convertir datos donde incident_count == 0 o null
df_clean['avg_resolution_time_hours'] = df_clean['avg_resolution_time_hours'].fillna(0)

# convertir datos donde sin_datos_monitoreo == True or null
monitoring_cols = [
    'avg_cpu_usage_pct', 'avg_memory_usage_pct', 'avg_response_time_ms',
    'avg_error_rate_pct', 'avg_availability_pct', 'avg_requests_per_minute'
]
for col in monitoring_cols:
    df_clean[col] = df_clean[col].fillna(0)

df_clean.to_csv('dataset_arreglado.csv', index=False)
print("Dataset arreglado guardado con éxito. Nulos restantes:")
print(df_clean.isnull().sum().sum())
```
