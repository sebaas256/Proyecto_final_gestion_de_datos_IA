```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path("/content/")

DATA_PROCESSED = BASE_DIR / "datos" / "procesados"
DATA_GEO = BASE_DIR / "datos" / "geo"

DATA_GEO.mkdir(parents=True, exist_ok=True)
DATA_PROCESSED.mkdir(parents=True, exist_ok=True)

print("carpeta geo:", DATA_GEO)
```

```python
#Cargamos dataset limpio
ruta_limpio = DATA_PROCESSED / "dataset_limpio.csv"

df = pd.read_csv(ruta_limpio)
print("Registros:", len(df))
print("Columnas:")
print(df.columns.tolist())

df.head()
```

```python
#verificamos el id de todas las empresas
empresas = sorted(df["company_id"].unique())
print("Cantidad de empresas:", len(empresas))
print("Empresas:")
print(empresas)
```

```python
#Ubicaciones con latitud y longitud
ubicaciones = {
    100001: ("San Salvador", "San Salvador", 13.6929, -89.2182),
    100002: ("La Libertad", "Santa Tecla", 13.6769, -89.2797),
    100003: ("Santa Ana", "Santa Ana", 13.9942, -89.5597),
    100004: ("San Miguel", "San Miguel", 13.4833, -88.1833),
    100005: ("La Paz", "Zacatecoluca", 13.5086, -88.8686),
    100006: ("Ahuachapán", "Ahuachapán", 13.9214, -89.8450),
    100007: ("Sonsonate", "Sonsonate", 13.7189, -89.7242),
    100008: ("Chalatenango", "Chalatenango", 14.0381, -88.9322),
    100009: ("Cuscatlán", "Cojutepeque", 13.7228, -88.9353),
    100010: ("Cabañas", "Sensuntepeque", 13.8761, -88.6331),
    100011: ("San Vicente", "San Vicente", 13.6442, -88.7778),
    100012: ("Usulután", "Usulután", 13.3444, -88.4392),
    100013: ("Morazán", "San Francisco Gotera", 13.6928, -88.1025),
    100014: ("La Unión", "La Unión", 13.3369, -87.8439),
    100015: ("San Salvador", "Soyapango", 13.7139, -89.1397),
    100016: ("San Salvador", "Mejicanos", 13.7314, -89.1914),
    100017: ("San Salvador", "Apopa", 13.8058, -89.1783),
    100018: ("La Libertad", "Antiguo Cuscatlán", 13.6733, -89.2536),
    100019: ("La Libertad", "Colón (Lourdes)", 13.7258, -89.3667),
    100020: ("Santa Ana", "Metapán", 14.3311, -89.4447),
    100021: ("Santa Ana", "Chalchuapa", 13.9856, -89.6806),
    100022: ("Sonsonate", "Izalco", 13.7431, -89.6731),
    100023: ("Usulután", "Jiquilisco", 13.3150, -88.5839),
    100024: ("San Miguel", "Ciudad Barrios", 13.7631, -88.2614),
    100025: ("La Libertad", "Puerto de La Libertad", 13.4883, -89.3200)
}
```

```python
#df geografico
registros_geo = []

for company_id, datos in ubicaciones.items():

    departamento, municipio, latitud, longitud = datos

    registros_geo.append({
        "company_id": company_id,
        "departamento": departamento,
        "municipio": municipio,
        "latitud": latitud,
        "longitud": longitud,
        "tipo_ubicacion": "Sintetica"
    })

df_geo = pd.DataFrame(registros_geo)

df_geo.head()
```

```python
#validacion de las empresas
print("Empresas en dataset:", df["company_id"].nunique())
print("Empresas en dimension geografica:", df_geo["company_id"].nunique())
faltantes_geo = set(df["company_id"]) - set(df_geo["company_id"])

print("Empresas sin ubicación:", faltantes_geo)
```

```python
#validacion de coordenadas
df_geo["latitud"] = pd.to_numeric(
    df_geo["latitud"],
    errors="coerce"
)

df_geo["longitud"] = pd.to_numeric(
    df_geo["longitud"],
    errors="coerce"
)
mask_valida = (
    df_geo["latitud"].between(13, 15) &
    df_geo["longitud"].between(-91, -87)
)

print("Total de empresas:", len(df_geo))
print("Coordenadas válidas:", mask_valida.sum())
print("Coordenadas a revisar:", (~mask_valida).sum())
```

```python
#guardamos la dimension geografica
ruta_geo = DATA_GEO / "empresas_geografia_sintetica.csv"

df_geo.to_csv(
    ruta_geo,
    index=False,
    encoding="utf-8"
)

print("Archivo guardado en:")
print(ruta_geo)
```

```python
#Unimos los datos por medio de la llave company_id
df_devops_geo = df.merge(
    df_geo,
    on="company_id",
    how="left"
)
print(df_devops_geo.shape)

df_devops_geo.head()
```

```python
#verificamos que no perdimos datos
print("Registros originales:", len(df))
print("Registros geográficos:", len(df_devops_geo))
```

```python
#guardamos el dataset listo para ArcGis
ruta_arcgis = DATA_GEO / "dataset_devops_geografico.csv"

df_devops_geo.to_csv(
    ruta_arcgis,
    index=False,
    encoding="utf-8"
)

print("Dataset para ArcGIS:")
print(ruta_arcgis)
```
